```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
dizi1 = np.linspace(1,6,10)
dizi2 = dizi1**2
```

```python
dizi1
```

```python
dizi2
```

```python
(figur,eksen) = plt.subplots()
eksen.plot(dizi1,dizi2,"black",alpha=0.5) #alpha= ile saydamlık ayarı yapabiliyoruz
eksen.plot(dizi2,dizi1,color="#48A3A8") #hex kodu olarak vermek istiyorsak eğer color= icine hex kodu yazılır
```

```python
from matplotlib.lines import lineStyles
(yeniFigur,yeniEksen) = plt.subplots()
yeniEksen.plot(dizi1,dizi2,"r")
yeniEksen.plot(dizi1,dizi2+2,"g")
yeniEksen.plot(dizi1,dizi2+5,"b",linewidth=5) #linewidth= ile çizgi kalınlığı değiştirilebilir
yeniEksen.plot(dizi1,dizi2+7,"black",linestyle="dotted") #linestyle ile çizginin şekli değiştirilebilir
yeniEksen.plot(dizi1,dizi2+9,color="purple",linestyle=":",linewidth=3,marker="o",markersize=6,markerfacecolor="red") #marker= ile verilerin bulundugu kısımları istedigimiz karakterler ile işaretleyebiliyoruz
```

# scatter

```python
plt.scatter(dizi1,dizi2) #grafiği nokta şeklinde gosterir
```
