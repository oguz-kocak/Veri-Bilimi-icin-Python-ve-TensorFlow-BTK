```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#%matplotlib.inline #grafik her seferinde gosterilir her seferinde plt.show() yazmamak icin kullanılır
```

```python
yasListesi = [1,2,3,4,5,6,7,8,9,5,3,22,32]
```

```python
kiloListesi = [4,2,7,3,2,6,9,16,35,22,1,3,4]
```

```python
numpyYasListesi = np.array(yasListesi)
```

```python
numpyKiloListesi = np.array(kiloListesi)
```

```python
numpyYasListesi
```

```python
numpyKiloListesi
```

```python
plt.plot(numpyYasListesi,numpyKiloListesi,"g") #ilk deger x ekseni , ikinci deger y ekseni , "g"green "r" red "b" blue
plt.xlabel("Yas") #X eksenine isim-etiket verir
plt.ylabel("Kilo") #Y eksenine isim-etiket verir
plt.title("Kilo'nun Yaş'a göre Değişim Grafiği") #grafiğe başlık verir
```

```python
dizi = np.linspace(0,10,20)
```

```python
dizi
```

```python
dizi2 = dizi**3
```

```python
dizi2
```

```python
plt.plot(dizi,dizi2,"y--") #y-- kesik kesik gorunum verir
```

```python
plt.plot(dizi2,dizi,"r*-") #y*- yildizlar ile gosterir
```

```python
plt.subplot(1,2,1) #subplot(satir,sutun,grafik numarası)
plt.plot(dizi,dizi2,"r*-")
plt.subplot(1,2,2) #1 sıra olacak , 2 kolon olacak , şu an ikinci grafiği çiziyorum
plt.plot(dizi2,dizi,"g--")
```

```python
from matplotlib import figure
benimFigure = plt.figure()
figureAxes = benimFigure.add_axes([0.2,0.2,0.9,0.9]) #[left,bottom,width,height]
figureAxes.plot(dizi,dizi2,"g")
figureAxes.set_xlabel("X ekseni veri ismi")
figureAxes.set_ylabel("Y ekseni veri ismi")
figureAxes.set_title("Grafik Başlığı")
```

bir figür içerisinde 2 grafik (**iç** içe grafik)

```python
figure2 = plt.figure()
eksen1 = figure2.add_axes([0.1,0.1,0.7,0.7])
eksen2 = figure2.add_axes([0.2,0.4,0.3,0.3])

eksen1.plot(dizi,dizi2,"r")
eksen1.set_xlabel("X ekseni")
eksen1.set_ylabel("Y ekseni")
eksen1.set_title("Ana grafik başlık")

eksen2.plot(dizi,dizi2,"g")
eksen2.set_xlabel("kucuk grafik X ekseni")
eksen2.set_ylabel("kucuk grafik Y ekseni")
eksen2.set_title("Kücük grafik başlık")
```

```python
figure , eksenler = plt.subplots(1,2)
type(eksenler)

for eksen in eksenler:
  eksen.plot(dizi,dizi2,"r")
  eksen.set_xlabel("X")
  eksen.set_ylabel("Y")
  eksen.set_title("title")

plt.tight_layout() #grafikler arasındaki boşlukları otomatik düzenler
```

```python
yeniFigur = plt.figure(dpi=90) #figure boyutunu figsize= ile belirleriz dpi= ile boyut ve kalite
yeniEksen = yeniFigur.add_axes([0.1,0.1,0.9,0.9])
yeniEksen.plot(dizi,dizi2,label = "dizi**2") #label grafikteki cizginin neyi ifade ettigini belirtmek icindir
yeniEksen.legend(loc=2) #bu isimleri grafik üzerinde gosterir loc= ile grafikteki yeri belirlenebilir
```

```python
yeniFigur.savefig("myfigure.png",dpi=200) #grafiğin resmini png olarak alırız
```
