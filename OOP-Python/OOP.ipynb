# inheritance

```python
class Hayvan():
  def __init__(self):
    print("hayvan sınıfı init çağrıldı")

  def method1(self):
    print("hayvan sınıfı method1 cagrıldi")

  def method2(self):
    print("hayvan sınıfı method 2 çagrıldı")
```

```python
benimHayvanim = Hayvan()
```

```python
benimHayvanim.method1()
```

```python
benimHayvanim.method2()
```

```python
class Kedi(Hayvan):
  def __init__(self):
    Hayvan.__init__(self)
    print("Kedi sınıfı çagrıldı")

  def miyavla(self):
    print("miyav")

  #Override
  def method1(self):
    print("method1 kedi sınıfından çagrıldı")
```

```python
benimKedi = Kedi()
```

```python
benimKedi.method1()
```

```python
benimKedi.miyavla()
```

```python
digerKedi = Kedi()
```

```python
digerKedi.method1() #Override ile çağrıldı
```

# Polymorphism

```python
class Elma():
  def __init__(self,isim):
    self.isim = isim

  def bilgiVer(self):
    return self.isim + " 100 kaloridir."
```

```python
class Muz:
  def __init__(self,isim):
    self.isim = isim

  def bilgiVer(self):
    return self.isim + " 150 kaloridir."
```

```python
elma = Elma("elma")
```

```python
elma.bilgiVer()
```

```python
muz = Muz("muz")
```

```python
muz.bilgiVer()
```

```python
meyveListesi = [elma,muz]
```

```python
for meyve in meyveListesi:
  print(meyve.bilgiVer())
```

```python
def bilgiAl(meyve):
  print(meyve.bilgiVer())
```

```python
bilgiAl(muz)
```

```python
bilgiAl(elma)
```
