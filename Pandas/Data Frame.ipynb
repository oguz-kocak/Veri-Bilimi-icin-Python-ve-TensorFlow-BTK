###Multi Index

```python
ilkIndexler = ["Simpson","Simpson","Simpson","South Park","South Park","South Park"]
```

```python
icIndexler = ["Homer","Bart","Marge","Cartman","Kenny","Kyle"]
```

```python
birlesmisIndex = list(zip(ilkIndexler,icIndexler))
```

```python
birlesmisIndex = pd.MultiIndex.from_tuples(birlesmisIndex) #Grup + alt grup ilişkilerinde kullanilir
```

```python
birlesmisIndex
```

```python
cizgiFilmListesi = [[40,"A"],[10,"B"],[30,"C"],[9,"D"],[10,"E"],[11,"F"]]
```

```python
cizgiFilmNumpyDizisi = np.array(cizgiFilmListesi)
```

```python
cizgiFilmDataFrame = pd.DataFrame(cizgiFilmNumpyDizisi,index=birlesmisIndex,columns=["Yas","Meslek"])
```

```python
cizgiFilmDataFrame
```
