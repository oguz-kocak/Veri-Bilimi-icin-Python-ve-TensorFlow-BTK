#Merge

```python
mergeSozluk1 = {"Isim" : ["Ahmet","Mehmet","Zeynep","Atil"],
                "Spor" : ["Kosu","Yuzme","Kosu","Basketbol"]}
```

```python
mergeDataFrame1 = pd.DataFrame(mergeSozluk1)
```

```python
mergeDataFrame1
```

```python
mergeSozluk2 = {"Isim" : ["Ahmet","Mehmet","Zeynep","Atil"],
                "Kalori" : [100,200,150,250]}
```

```python
mergeDataFrame2 = pd.DataFrame(mergeSozluk2)
```

```python
mergeDataFrame2
```

```python
pd.merge(mergeDataFrame1,mergeDataFrame2,on = "Isim") #ortak kolona gore eslestir (on ile ortak kolon adini belirtiriz)
```

```python
maasSozluk = {"Isimler" : ["Atil","Zeynep","Mehmet","Ahmet"],
              "Departman" : ["Yazilim","Satis","Pazarlama","Yazilim"],
              "Maas" : [200,300,400,500]}
```

```python
maasDataFrame = pd.DataFrame(maasSozluk)
```

```python
maasDataFrame
```

```python
maasDataFrame["Departman"].unique() #sutundaki degerleri getirir (tekrar edenleri bir kere getirir)
```

```python
maasDataFrame["Departman"].nunique() #number of unique kac tane departman oldugunu bulur
```

```python
maasDataFrame["Departman"].value_counts() #hangi departmandan kac tane kisi oldugunu getirir
```

```python
def bruttenNete(maas):
  return maas * 0.66
```

```python
maasDataFrame["Maas"].apply(bruttenNete) #her maas degerine bruttenNete fonksiyonunu uygula (0*66 ile carp)
```

```python
maasDataFrame.isnull() #bos deger var mı ? (true = bos deger) (false = bos degil)
```
