###Ders 1 Seriler

```python
import numpy as np
import pandas as pd
```

```python
sozluk = {"oguz":1 ,"ali":2 ,"zeynep":3}
```

```python
pd.Series(sozluk)
```

```python
yaslar = [50,40,30]
```

```python
isimler = ["oguz","ali","zeynep"]
```

```python
pd.Series(yaslar)
```

```python
pd.Series(yaslar,isimler)
```

```python
pd.Series(data = yaslar,index = isimler)
```

```python
numpyDizisi = np.array([50,40,30])
```

```python
numpyDizisi
```

```python
pd.Series(numpyDizisi)
```

```python
pd.Series(numpyDizisi,isimler)
```

###Ders 2 Series Ozellikleri

```python
pd.Series(["Atil","Atlas","Osman"],[1,2,3]) #indexleri degistirdik normalde indexler 0,1,2 seklinde
```

```python
yarismaSonucu = pd.Series([10,5,1],["Atil","Atlas","Osman"])
```

```python
yarismaSonucu
```

```python
yarismaSonucu2 = pd.Series([20,10,8],["Atil","Atlas","Osman"])
```

```python
yarismaSonucu2
```

```python
sonSonuc = yarismaSonucu + yarismaSonucu2
```

```python
sonSonuc
```

```python
farkliSeries = pd.Series([20,30,40,50],["a","b","c","d"])
```

```python
farkliSeries2 = pd.Series([10,5,3,1],["a","c","f","g"])
```

```python
farkliSeries + farkliSeries2 #ortak olan degerleri toplar (kesisenleri toplar)
```
