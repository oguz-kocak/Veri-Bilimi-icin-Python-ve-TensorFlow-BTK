# Yeni Ornek

```python
yeniVeri = {"Istanbul":[30,29,np.nan] , "Ankara":[20,np.nan,25] , "Izmır":[40,39,38] , "Antalya":[45,np.nan,np.nan]} #np.nan = not a number (veri yok)
yeniDataFrame = pd.DataFrame(yeniVeri)
```

```python
yeniDataFrame
```

```python
yeniDataFrame.dropna(axis=1,thresh=2) #sutunlarda axis=1 , en az 2 tane NaN olmayan (dolu) varsa o satir kalir digerleri silinir thresh=2
```

```python
yeniDataFrame.fillna(20) #NaN olan yerlerin hepsine 20 degerini koy
```

#groupby

```python
maasSozlugu = {"Departman" : ["Yazilim","Yazilim","Pazarlama","Pazarlama","Hukuk","Hukuk"],
               "Calisan Ismi" : ["Ahmet","Mehmet","Atil","Burak","Zeynep","Fatma"],
               "Maas" : [100,150,200,300,400,500]}
```

```python
maasDataFrame = pd.DataFrame(maasSozlugu)
```

```python
maasDataFrame
```

```python
grupObjesi = maasDataFrame.groupby("Departman") #her grup uzerinde islem yapmamizi saglar
```

```python
grupObjesi.count() #kac tane var
```

```python
grupObjesi["Maas"].mean() #her bir Departmanın ortalamasini bulur
```

```python
grupObjesi.min() #her bir Departmanın en dusuk Maasli verisi
```

```python
grupObjesi.max() #her bir Departmanın en yuksek Maasli verisi
```

```python
grupObjesi.describe() #her bir departmana ait bazi ozellikleri tek bir tabloda hesapliyor (std = standart sapma)
```
